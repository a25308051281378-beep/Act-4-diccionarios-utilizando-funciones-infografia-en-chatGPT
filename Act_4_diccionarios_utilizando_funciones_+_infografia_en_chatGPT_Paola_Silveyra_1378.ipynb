#### Paola Silveyra NL = 55 Fecha 18/05/26

#### Análisis del problema
#### Se trabajará con un diccionario vacío llamado diccionario_modelos.
#### Cada modelo tendrá 6 atributos:

#### Agencia
* Estatura
* Tipo de modelaje
* Marcas representadas
* Portafolio
* Disponibilidad de viaje

#### Funciones requeridas:

* agregar_modelo()
* listar_modelos()
* buscar_modelo()
* borrar_modelo()
* actualizar_modelo()

#### Se utilizará:

* while True
* if elif
* for clave, valor in
* Menú interactivo
* Tablas para mostrar información

```python
# ==========================================
# ACT 4 - DICCIONARIOS UTILIZANDO FUNCIONES
# EJERCICIO 55 - MODELO
# ==========================================

diccionario_modelos = {}


def listar_modelos():

    print("\n" + "=" * 95)
    print(f"{'NOMBRE':15} {'AGENCIA':15} {'ESTATURA':10} {'TIPO':15} {'MARCAS':15} {'PORTAFOLIO':12} {'VIAJE':10}")
    print("=" * 95)

    for nombre, datos in diccionario_modelos.items():

        print(
            f"{nombre:15} "
            f"{datos['Agencia']:15} "
            f"{datos['Estatura']:10} "
            f"{datos['Tipo']:15} "
            f"{datos['Marcas']:15} "
            f"{datos['Portafolio']:12} "
            f"{datos['Viaje']:10}"
        )

    print("=" * 95)


def agregar_modelo():

    nombre = input("Nombre de la modelo: ")

    agencia = input("Agencia: ")
    estatura = input("Estatura: ")
    tipo = input("Tipo de modelaje: ")
    marcas = input("Marcas representadas: ")
    portafolio = input("Portafolio: ")
    viaje = input("Disponibilidad de viaje: ")

    diccionario_modelos[nombre] = {
        "Agencia": agencia,
        "Estatura": estatura,
        "Tipo": tipo,
        "Marcas": marcas,
        "Portafolio": portafolio,
        "Viaje": viaje
    }

    print("\nModelo agregada correctamente.")
    listar_modelos()


def buscar_modelo():

    nombre = input("Nombre de la modelo a buscar: ")

    if nombre in diccionario_modelos:

        print("\nInformación encontrada:\n")

        for clave, valor in diccionario_modelos[nombre].items():
            print(clave, ":", valor)

    else:
        print("Modelo no encontrada.")

    listar_modelos()


def borrar_modelo():

    nombre = input("Nombre de la modelo a borrar: ")

    if nombre in diccionario_modelos:
        del diccionario_modelos[nombre]
        print("Modelo eliminada correctamente.")
    else:
        print("Modelo no encontrada.")

    listar_modelos()


def actualizar_modelo():

    nombre = input("Nombre de la modelo a actualizar: ")

    if nombre in diccionario_modelos:

        agencia = input("Nueva agencia: ")
        estatura = input("Nueva estatura: ")
        tipo = input("Nuevo tipo de modelaje: ")
        marcas = input("Nuevas marcas representadas: ")
        portafolio = input("Nuevo portafolio: ")
        viaje = input("Nueva disponibilidad de viaje: ")

        diccionario_modelos[nombre] = {
            "Agencia": agencia,
            "Estatura": estatura,
            "Tipo": tipo,
            "Marcas": marcas,
            "Portafolio": portafolio,
            "Viaje": viaje
        }

        print("Modelo actualizada correctamente.")

    else:
        print("Modelo no encontrada.")

    listar_modelos()


while True:

    print("\n===== MENÚ MODELOS =====")
    print("1. Agregar modelo")
    print("2. Listar modelos")
    print("3. Buscar modelo")
    print("4. Borrar modelo")
    print("5. Actualizar modelo")
    print("6. Salir")

    opcion = input("Seleccione una opción: ")

    if opcion == "1":
        agregar_modelo()

    elif opcion == "2":
        listar_modelos()

    elif opcion == "3":
        buscar_modelo()

    elif opcion == "4":
        borrar_modelo()

    elif opcion == "5":
        actualizar_modelo()

    elif opcion == "6":
        print("Fin del programa.")
        break

    else:
        print("Opción no válida.")
```
